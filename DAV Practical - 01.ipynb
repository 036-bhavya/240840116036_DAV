**Aim : Study the Data Analytics lifecycle and perform data acquisition from CSV, Excel, and JSON ƒles using Pandas.**

**Step - 01 : Import the pandas library.**

```python
#First import the pandas to perform Data Analytics

import pandas as pd
```

**Step - 02 : Create sample data.**

```python
#Get/Create a Data on which we will perform practical.

data = {
                "ID" : [101, 102, 103, 104, 105, 106],
                "Name" : ["Rahul", "Priya", "Rajan", "Krisha", "Ruchi", "Dhruvil"],
                "Age" : [20, 21, 13, 20, 22, 20],
                "City" : ["Surat", "Ahemdabad", "Surat", "Vadodara", "Gandhinagar", "Surat"],
                "Marks" : [89, 95, 75, 84, 96, 74]
}
```

**Step - 03 : Create a Dataframe.**

```python
#Create a DataFrame to organise the data.

df = pd.DataFrame(data)

df
```

**Step - 04 : Save Data as a CSV File.**

```python
#Save the data in csv file format.

df.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

**Step - 05 : Save Data as an Excel File.**

```python
#Save data in excel file format.

df.to_excel("students.xlsx", index=False)

print("Excel file is created successfully.")
```

**Step - 06 : Save Data as a JSON File.**

```python
#Save data in JSON file format.

df.to_json("students.json", orient="records", index=4)

print("JSON file is created successfully.")
```
